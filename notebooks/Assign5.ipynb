```python
import pandas as pd
#change each file from CSV to xlsx

df = pd.read_csv("data/aos-bordercounties_-_01.10.2020.csv")
df.to_excel("borderStates.xlsx", index=False)

df = pd.read_csv("data/scholars_unders_surveillance_dataset_03-03-2021.csv")
df.to_excel("colleges.xlsx", index=False)

df = pd.read_csv("data/EFF-MuckRock 2016-2017 ALPR DATA.csv")
df.to_excel("alpr.xlsx", index=False)

df = pd.read_csv("data/Atlas of Surveillance-20251114.csv")
df.to_excel("allStates.xlsx", index=False)

```

```python
#opens the xlsx file 
df1 = pd.read_excel("borderStates.xlsx")
df2 = pd.read_excel("colleges.xlsx")
df3 = pd.read_excel("alpr.xlsx")
df4 = pd.read_excel("allStates.xlsx")
```

```python
import sqlite3

#creates a database with tables based on our xlsx files
conn = sqlite3.connect("mydatabase.db")

df1.to_sql("borderStates", conn, if_exists="replace", index=False)
df2.to_sql("college", conn, if_exists="replace", index=False)
df3.to_sql("alpr", conn, if_exists="replace", index=False)
df4.to_sql("aos", conn, if_exists="replace", index=False)

```

```python
###Our sql version of our 'Technology count for All states' Graph 

sql = "SELECT Technology, COUNT(*) AS Count FROM aos GROUP BY Technology;"

pd.read_sql_query(sql, conn)
```

```python
#sql version of our 'Technology count from the Police from all states' graph
sql = "SELECT Technology, COUNT(*) AS Count FROM aos WHERE TypeOfLEA = 'Police' GROUP BY Technology;"

pd.read_sql_query(sql, conn)
```

```python
###our sql version of our 'Technology count by Type of Lea' 

#cleans and trims our LEA column 
conn.execute("UPDATE college SET TypeOfLEA = LOWER(TRIM(TypeOfLEA));")
conn.commit()


#groups the data by LEA and technoloy, retrives only the technology, type of LEA, and counts of each 
sql = """ SELECT TypeOfLEA, Technology, COUNT(*) AS count
FROM college
GROUP BY TypeOfLEA, Technology;
"""

pd.read_sql_query(sql, conn)

```

```python
###Our sql version of our 'Technology count for Border States' Graph 

sql = "SELECT TechnologyType, COUNT(*) AS Count FROM borderStates GROUP BY TechnologyType;"

pd.read_sql_query(sql, conn)
```

```python
###sql version of our 'Technology Type Used by Police in Border States' graph
#our grouped border states grouped by only police, showing the counts of each technology type

sql = "SELECT TechnologyType, COUNT(*) AS Count FROM borderStates WHERE LEAType = 'Police' GROUP BY TechnologyType;"

pd.read_sql_query(sql, conn)
```

```python
#to rename the 2016To2017Detections column so that we can access the column in sql without problems 
conn.execute('ALTER TABLE alpr RENAME COLUMN "2016To2017Detections" TO detections_2016_2017;')
conn.commit()
```

```python
#to rename the 2016To2017Hits column so that we can access the column in sql without problems 
conn.execute('ALTER TABLE alpr RENAME COLUMN "2016To2017Hits" TO hits_2016_2017;')
conn.commit()
```

```python
#to check if renaming the column worked/ went through 
pd.read_sql_query("PRAGMA table_info(alpr);", conn)
```

```python
#the sql version of our 

#cleans the data, gets rid of any spaces, dashes, commas, etc from both of the 2016to2017 hits and detections columns
conn.execute("UPDATE alpr SET detections_2016_2017 = REPLACE(REPLACE(REPLACE(detections_2016_2017, ',', ''), ' ', ''), '-', ''), hits_2016_2017       = REPLACE(REPLACE(REPLACE(hits_2016_2017, ',', ''), ' ', ''), '-', '');")
conn.commit()

#converts any values into integer values 
conn.execute("UPDATE alpr SET detections_2016_2017 = CAST(detections_2016_2017 AS INTEGER),hits_2016_2017       = CAST(hits_2016_2017 AS INTEGER);")
conn.commit()

#selects the sumed up ALPR detections from 2016 to 2017 by state, sorted alphabetically 
sql = """SELECT State, SUM(detections_2016_2017) AS TotalDetections
FROM alpr
GROUP BY State
ORDER BY State;
"""

pd.read_sql_query(sql, conn)
```

```python
#the sql version of the total ALPR hits from 2016 to 2017, grouped by state and ordered alphabetically 
sql = """SELECT State, SUM(hits_2016_2017) AS TotalHits
FROM alpr
GROUP BY State
ORDER BY State;
"""

pd.read_sql_query(sql, conn)
```
